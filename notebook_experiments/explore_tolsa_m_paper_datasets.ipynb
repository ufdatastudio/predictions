```python
import re

# modal verbs
modal_words = [
    'will', 'would', 'may', 'might', 'could',
    'should', 'shall', 'must', 'can'
]

# predictive / forward-looking terms
predictive_words = [
    'predict', 'predicts', 'predicted', 'predicting',
    'forecast', 'forecasts', 'forecasted', 'forecasting',
    'expect', 'expects', 'expected', 'expecting',
    'project', 'projects', 'projected', 'projecting',
    'anticipate', 'anticipates', 'anticipated', 'anticipating',
    'speculate', 'speculates', 'speculated', 'speculating',
    'envision', 'envisions', 'envisioned', 'envisioning',
    'likely', 'probable', 'probably'
]

exclude_words = modal_words + predictive_words

pattern = r'\b(?:' + '|'.join(map(re.escape, exclude_words)) + r')\b'

tolsa_m = df[df['Ground Truth'] == 1]

no_modal_no_predictive = tolsa_m[
    ~tolsa_m['Base Sentence']
      .str.contains(pattern, case=False, na=False, regex=True)
]

print(no_modal_no_predictive.shape)
display(no_modal_no_predictive[['Base Sentence']].head(20))
```
