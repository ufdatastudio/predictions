## Keywords

```python
def load_baseline_results(model_type, result_dir):
    records = []
    embedding_models = (
        ["spacy_small", "spacy_medium", "spacy_large"]
        if model_type == "glodd"
        else [None]
    )

    for seed in seeds:
        seed_dir = result_dir / f"seed{seed}"

        for embedding in embedding_models:
            if model_type == "glodd":
                search_dir = (
                    seed_dir / "in_domain"
                    / "glodd_hristova_step0" / embedding
                )
                target_file = "metrics_summary_glodd_step0.csv"
            elif model_type == "conll":
                search_dir = (
                    seed_dir / "in_domain" / "conll_2010_hedge_rule"
                )
                target_file = "metrics_summary_conll.csv"
            else:
                raise ValueError(f"Unsupported model type: {model_type}")

            if not search_dir.exists():
                print(f"Missing directory: {search_dir}")
                continue

            for metrics_file in search_dir.rglob(target_file):
                df = pd.read_csv(metrics_file)
                if df.empty:
                    continue

                for _, row in df.iterrows():
                    record = {
                        "seed": seed,
                        "model_type": model_type,
                        "model": embedding if model_type == "glodd" else row.get("model", model_type),
                    }

                    if embedding is not None:
                        record["embedding_model"] = embedding

                    for metric in [
                        "val_accuracy", "test_accuracy",
                        "val_precision_class_1", "val_recall_class_1",
                        "val_f1_class_1", "test_precision_class_1",
                        "test_recall_class_1", "test_f1_class_1",
                    ]:
                        if metric in df.columns:
                            record[metric] = pd.to_numeric(
                                row[metric], errors="coerce"
                            )

                    records.append(record)

    raw_df = pd.DataFrame(records)

    if raw_df.empty:
        print(f"No {model_type} results found.")
        return raw_df, raw_df, raw_df

    group_cols = ["model_type", "model"]
    if "embedding_model" in raw_df.columns:
        group_cols.append("embedding_model")

    metric_cols = [
        col for col in raw_df.select_dtypes(include=np.number).columns
        if col != "seed"
    ]

    summary_df = (
        raw_df.groupby(group_cols, as_index=False)
        .agg(
            **{metric: (metric, "mean") for metric in metric_cols},
            n_seeds=("seed", "nunique"),
        )
    )

    std_df = (
        raw_df.groupby(group_cols)[metric_cols]
        .std()
        .reset_index()
    )

    return raw_df, summary_df, std_df


glodd_raw, glodd_summary, glodd_std = load_baseline_results(
    "glodd", experiment_dir
)

conll_raw, conll_summary, conll_std = load_baseline_results(
    "conll", experiment_dir
)

display(glodd_raw)
display(glodd_summary)
display(conll_raw)
display(conll_summary)

# avg across seeds, so 
```

## LLMs

```python
models = [
    "gpt-oss-20b",
    "llama-3.3-70b-instruct",
    "mistral-small-3.1",
]

model_labels = {
    "gpt-oss-20b": "GPT-OSS-20B",
    "llama-3.3-70b-instruct": "Llama-3.3-70B",
    "mistral-small-3.1": "Mistral-Small-3.1",
}

prompt_labels = {
    "zero-shot": "ZS",
    "chain-of-thought": "CoT",
}

llm_records = []

for seed in seeds:
    llm_dir = (
        experiment_dir / f"seed{seed}" / "in_domain" / "llm"
    )

    for model in models:
        model_dir = llm_dir / model

        if not model_dir.exists():
            continue

        for prompt in prompt_labels:
            metrics_file = (
                model_dir / prompt / "test"
                / f"metrics_summary_{model}_{prompt}.csv"
            )

            if not metrics_file.exists():
                continue

            df = pd.read_csv(metrics_file)

            for _, row in df.iterrows():
                llm_records.append({
                    "seed": seed,
                    "model": model,
                    "prompt_type": prompt,
                    "test_accuracy": row.get("test_accuracy", np.nan),
                    "macro_f1": np.mean([
                        row.get("f1_class_0", np.nan),
                        row.get("f1_class_1", np.nan),
                    ]),
                })

llm_seed_results = pd.DataFrame(llm_records)

llm_summary = (
    llm_seed_results
    .groupby(["model", "prompt_type"], as_index=False)
    .agg(
        test_accuracy=("test_accuracy", "mean"),
        test_accuracy_std=("test_accuracy", "std"),
        macro_f1=("macro_f1", "mean"),
        macro_f1_std=("macro_f1", "std"),
        n_seeds=("seed", "nunique"),
    )
)

llm_summary["model_label"] = llm_summary["model"].map(model_labels)
llm_summary["prompt_label"] = llm_summary["prompt_type"].map(prompt_labels)

display(llm_seed_results)
display(llm_summary)
```

```python
# =====================================================
# ML Model Results
# =====================================================

embedding_models = {
    "spacy_small": "spaCy-S",
    "spacy_medium": "spaCy-M",
    "spacy_large": "spaCy-L",
    "st_distilroberta": "ST-DR",
    "st_minilm_l12": "ST-ML12",
    "st_minilm_l6": "ST-ML6",
    "st_mpnet_base": "ST-MPNet",
}

ml_records = []

for seed in seeds:
    seed_dir = experiment_dir / f"seed{seed}" / "in_domain"

    for embedding, model_label in embedding_models.items():
        metrics_file = (
            seed_dir / embedding / "metrics_summary_ml_models.csv"
        )

        if not metrics_file.exists():
            print(f"Missing: {metrics_file}")
            continue

        df = pd.read_csv(metrics_file)

        if df.empty:
            continue

        for _, row in df.iterrows():
            f1_0 = pd.to_numeric(row.get("f1_class_0", np.nan), errors="coerce")
            f1_1 = pd.to_numeric(row.get("f1_class_1", np.nan), errors="coerce")

            ml_records.append({
                "seed": seed,
                "model": embedding,
                "model_label": model_label,
                "test_accuracy": pd.to_numeric(
                    row.get("test_accuracy", np.nan), errors="coerce"
                ),
                "macro_f1": np.mean([f1_0, f1_1]),
            })

ml_seed_results = pd.DataFrame(ml_records)

ml_summary = (
    ml_seed_results
    .groupby(["model", "model_label"], as_index=False)
    .agg(
        test_accuracy=("test_accuracy", "mean"),
        test_accuracy_std=("test_accuracy", "std"),
        macro_f1=("macro_f1", "mean"),
        macro_f1_std=("macro_f1", "std"),
        n_seeds=("seed", "nunique"),
    )
)

display(ml_seed_results)
display(ml_summary)
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from matplotlib.patches import Patch

# =====================================================
# Build Individual Model Results from Existing DataFrames
# =====================================================
plot_records = []

# =====================================================
# Baseline Keywords (Glodd)
# =====================================================
glodd_labels = {
    "spacy_small": "Glodd-S",
    "spacy_medium": "Glodd-M",
    "spacy_large": "Glodd-L",
}

for _, row in glodd_summary.iterrows():
    embedding = row["embedding_model"]
    std_row = glodd_std[
        glodd_std["embedding_model"] == embedding
    ]

    if {
        "test_f1_class_0",
        "test_f1_class_1",
    }.issubset(glodd_summary.columns):

        f1_value = np.mean([
            row["test_f1_class_0"],
            row["test_f1_class_1"],
        ])

        f1_sd = (
            np.sqrt(
                std_row["test_f1_class_0"].iloc[0] ** 2
                + std_row["test_f1_class_1"].iloc[0] ** 2
            ) / 2
            if not std_row.empty
            else np.nan
        )

    else:
        f1_value = row.get("test_f1_class_1", np.nan)

        f1_sd = (
            std_row["test_f1_class_1"].iloc[0]
            if (
                not std_row.empty
                and "test_f1_class_1" in std_row.columns
            )
            else np.nan
        )

    plot_records.append({
        "Model": glodd_labels[embedding],
        "Accuracy": row.get("test_accuracy", np.nan),
        "Accuracy SD": (
            std_row["test_accuracy"].iloc[0]
            if not std_row.empty
            else np.nan
        ),
        "Macro F1": f1_value,
        "Macro F1 SD": f1_sd,
    })

# =====================================================
# Baseline Keywords (CoNLL10)
# =====================================================
for _, row in conll_summary.iterrows():
    model_value = row.get("model", "conll")
    std_row = conll_std[
        conll_std["model"] == model_value
    ]

    if {
        "test_f1_class_0",
        "test_f1_class_1",
    }.issubset(conll_summary.columns):

        f1_value = np.mean([
            row["test_f1_class_0"],
            row["test_f1_class_1"],
        ])

        f1_sd = (
            np.sqrt(
                std_row["test_f1_class_0"].iloc[0] ** 2
                + std_row["test_f1_class_1"].iloc[0] ** 2
            ) / 2
            if not std_row.empty
            else np.nan
        )

    else:
        f1_value = row.get("test_f1_class_1", np.nan)

        f1_sd = (
            std_row["test_f1_class_1"].iloc[0]
            if (
                not std_row.empty
                and "test_f1_class_1" in std_row.columns
            )
            else np.nan
        )

    plot_records.append({
        "Model": "CoNLL10",
        "Accuracy": row.get("test_accuracy", np.nan),
        "Accuracy SD": (
            std_row["test_accuracy"].iloc[0]
            if not std_row.empty
            else np.nan
        ),
        "Macro F1": f1_value,
        "Macro F1 SD": f1_sd,
    })

# =====================================================
# Classical ML
# =====================================================
ml_labels = {
    "spacy_small": "spaCy-S",
    "spacy_medium": "spaCy-M",
    "spacy_large": "spaCy-L",
    "st_distilroberta": "ST-DR",
    "st_minilm_l12": "ST-ML12",
    "st_minilm_l6": "ST-ML6",
    "st_mpnet_base": "ST-MPNet",
}

for _, row in ml_summary.iterrows():
    model_key = row["model"]

    plot_records.append({
        "Model": ml_labels.get(
            model_key,
            row.get("model_label", model_key),
        ),
        "Accuracy": row.get("test_accuracy", np.nan),
        "Accuracy SD": row.get("test_accuracy_std", np.nan),
        "Macro F1": row.get("macro_f1", np.nan),
        "Macro F1 SD": row.get("macro_f1_std", np.nan),
    })

# =====================================================
# Prompted LLMs: Individual Zero-Shot Models Only
# =====================================================
llm_models = [
    "mistral-small-3.1",
    "gpt-oss-20b",
    "llama-3.3-70b-instruct",
]

llm_labels = {
    "mistral-small-3.1": "Mistral-Small-3.1 (ZS)",
    "gpt-oss-20b": "GPT-OSS-20B (ZS)",
    "llama-3.3-70b-instruct": "Llama-3.3-70B (ZS)",
}

llm_plot = (
    llm_seed_results[
        (llm_seed_results["prompt_type"] == "zero-shot")
        & (llm_seed_results["model"].isin(llm_models))
    ]
    .groupby("model", as_index=False)
    .agg(
        Accuracy=("test_accuracy", "mean"),
        Accuracy_SD=("test_accuracy", "std"),
        Macro_F1=("macro_f1", "mean"),
        Macro_F1_SD=("macro_f1", "std"),
    )
)

for _, row in llm_plot.iterrows():
    plot_records.append({
        "Model": llm_labels[row["model"]],
        "Accuracy": row["Accuracy"],
        "Accuracy SD": row["Accuracy_SD"],
        "Macro F1": row["Macro_F1"],
        "Macro F1 SD": row["Macro_F1_SD"],
    })

# =====================================================
# TOLSA-BERT-Base and TOLSA-BERT-Large
# Requires the previously loaded bert_comparison DataFrame.
# =====================================================
def get_bert_metric(row, metric, statistic):
    candidates = {
        ("accuracy", "mean"): [
            "test_accuracy_mean",
            "test_accuracy",
            "mean_test_accuracy",
        ],
        ("accuracy", "std"): [
            "test_accuracy_std",
            "std_test_accuracy",
        ],
        ("f1", "mean"): [
            "test_f1_class_1_mean",
            "test_f1_class_1",
        ],
        ("f1", "std"): [
            "test_f1_class_1_std",
        ],
    }

    for column in candidates[(metric, statistic)]:
        if (
            column in row.index
            and pd.notna(row[column])
        ):
            return pd.to_numeric(
                row[column],
                errors="coerce",
            )

    return np.nan


if (
    "bert_comparison" in globals()
    and not bert_comparison.empty
):
    for embedding, label in [
        ("BERT-Base", "TOLSA-BERT-Base"),
        ("BERT-Large", "TOLSA-BERT-Large"),
    ]:
        bert_row = bert_comparison[
            bert_comparison["Embedding"] == embedding
        ]

        if bert_row.empty:
            print(
                f"Warning: {embedding} not found "
                "in bert_comparison."
            )
            continue

        row = bert_row.iloc[0]

        plot_records.append({
            "Model": label,
            "Accuracy": get_bert_metric(
                row, "accuracy", "mean"
            ),
            "Accuracy SD": get_bert_metric(
                row, "accuracy", "std"
            ),
            "Macro F1": get_bert_metric(
                row, "f1", "mean"
            ),
            "Macro F1 SD": get_bert_metric(
                row, "f1", "std"
            ),
        })

    print(
        "BERT columns:",
        bert_comparison.columns.tolist(),
    )

else:
    print(
        "Warning: bert_comparison is missing or empty. "
        "Load the BERT CSVs first."
    )

# =====================================================
# Combine Results
# =====================================================
plot_df = (
    pd.DataFrame(plot_records)
    .drop_duplicates("Model")
    .set_index("Model")
)

# =====================================================
# Plot Order
# =====================================================
models = [
    "Glodd-S",
    "Glodd-M",
    "Glodd-L",
    "CoNLL10",
    "spaCy-S",
    "spaCy-M",
    "spaCy-L",
    "ST-DR",
    "ST-ML12",
    "ST-ML6",
    "ST-MPNet",
    "Mistral-Small-3.1 (ZS)",
    "GPT-OSS-20B (ZS)",
    "Llama-3.3-70B (ZS)",
    "TOLSA-BERT-Base",
    "TOLSA-BERT-Large",
]

plot_df = plot_df.reindex(models)

acc = plot_df["Accuracy"].to_numpy(dtype=float)
acc_std = plot_df["Accuracy SD"].to_numpy(dtype=float)
f1 = plot_df["Macro F1"].to_numpy(dtype=float)
f1_std = plot_df["Macro F1 SD"].to_numpy(dtype=float)

# =====================================================
# Colors
# =====================================================
# Baseline Keywords: Glodd + CoNLL10
# Classical ML: spaCy features
# Classical ML: Sentence Transformer features
# Prompted LLMs: individual zero-shot models
# TOLSA-BERT: Base + Large

colors = (
    ["#DD8452"] * 4
    + ["#4C72B0"] * 3
    + ["#55A868"] * 4
    + ["#C44E52"] * 3
    + ["#8172B2"] * 2
)

x = np.arange(len(models))

# =====================================================
# Broken Axis Configuration
# =====================================================
GAP = 0.08


def add_y_axis_break(
    ax,
    y0,
    y1,
    width=0.006,
    n_zigs=3,
):
    ax.plot(
        [0, 0],
        [y0, y1],
        transform=ax.transAxes,
        color="white",
        linewidth=4,
        clip_on=False,
        zorder=10,
    )

    ys = np.linspace(y0, y1, n_zigs + 2)

    xs = [0] + [
        -width if i % 2 == 0 else width
        for i in range(n_zigs)
    ] + [0]

    ax.plot(
        xs,
        ys,
        transform=ax.transAxes,
        color="black",
        linewidth=1,
        clip_on=False,
        zorder=11,
    )


# =====================================================
# Plot Helper
# =====================================================
def plot_broken_panel(
    ax,
    values,
    stds,
    lower,
    ylabel,
    title,
    upper=1.0,
):
    values = np.asarray(values, dtype=float)
    stds = np.asarray(stds, dtype=float)

    valid = np.isfinite(values)

    heights = GAP + (values - lower)

    bars = ax.bar(
        x[valid],
        heights[valid],
        yerr=np.where(
            np.isfinite(stds[valid]),
            stds[valid],
            0,
        ),
        color=np.asarray(colors)[valid],
        edgecolor="black",
        linewidth=0.8,
        capsize=4,
    )

    top = GAP + (upper - lower)
    ax.set_ylim(0, top)

    ticks = np.round(
        np.arange(lower, upper + 1e-9, 0.1),
        1,
    )

    ax.set_yticks(
        [0] + list(GAP + (ticks - lower))
    )

    ax.set_yticklabels(
        ["0"] + [f"{t:.1f}" for t in ticks]
    )

    ax.set_ylabel(ylabel, fontsize=11)
    ax.set_title(title, fontsize=12, weight="bold")
    ax.grid(axis="y", linestyle="--", alpha=0.35)
    ax.set_axisbelow(True)

    for bar, value in zip(bars, values[valid]):
        ax.text(
            bar.get_x() + bar.get_width() / 2,
            bar.get_height() + 0.01,
            f"{value:.3f}",
            ha="center",
            va="bottom",
            fontsize=7,
            rotation=90,
        )

    add_y_axis_break(
        ax,
        y0=0,
        y1=GAP / top,
    )


# =====================================================
# Create Figure
# =====================================================
fig, axes = plt.subplots(
    2,
    1,
    figsize=(18, 9),
    sharex=True,
)

plot_broken_panel(
    axes[0],
    acc,
    acc_std,
    lower=0.60,
    ylabel="Test Accuracy",
    title="Test Accuracy by Model",
)

plot_broken_panel(
    axes[1],
    f1,
    f1_std,
    lower=0.40,
    ylabel="Macro F1",
    title="Macro F1 by Model",
)

axes[1].set_xticks(x)
axes[1].set_xticklabels(
    models,
    rotation=45,
    ha="right",
    fontsize=9,
)

# =====================================================
# Legend
# =====================================================
legend_handles = [
    Patch(
        facecolor="#DD8452",
        edgecolor="black",
        label="Baseline Keywords",
    ),
    Patch(
        facecolor="#4C72B0",
        edgecolor="black",
        label="Classical ML (spaCy Features)",
    ),
    Patch(
        facecolor="#55A868",
        edgecolor="black",
        label="Classical ML (Sentence Transformers)",
    ),
    Patch(
        facecolor="#C44E52",
        edgecolor="black",
        label="Prompted LLMs (Zero-Shot)",
    ),
    Patch(
        facecolor="#8172B2",
        edgecolor="black",
        label="TOLSA-BERT",
    ),
]

fig.legend(
    handles=legend_handles,
    loc="upper center",
    bbox_to_anchor=(0.5, 1.03),
    ncol=3,
    frameon=False,
    fontsize=9,
)

plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()
```

## Glodd Keywords

```python
from pathlib import Path
import pandas as pd

base_dir = Path(
    "../data/classification_results/naacl_2026_submission/"
    "tolsa_naacl_2026_2027-final"
)

seeds = [3, 7, 33]
embeddings = ["spacy_small", "spacy_medium", "spacy_large"]
distribution_files = [
    "distribution_all.csv",
    "distribution_tolsam.csv",
    "distribution_non_tolsam.csv",
]

glodd_results = {}

for seed in seeds:
    for embedding in embeddings:
        seed_dir = (
            base_dir
            / f"seed{seed}"
            / "in_domain"
            / "glodd_hristova_step0"
            / embedding
        )

        for filename in distribution_files:
            filepath = seed_dir / filename
            key = f"{filename.removesuffix('.csv')}_seed{seed}_{embedding}"

            if filepath.exists():
                glodd_results[key] = pd.read_csv(filepath)
                print(f"Loaded {key}: {glodd_results[key].shape}")
            else:
                print(f"Missing: {filepath}")

globals().update(glodd_results)
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns

SEEDS = [3, 7, 33]
EMBEDDINGS = ["spacy_small", "spacy_medium", "spacy_large"]
LABEL_COLUMN = "Ground Truth"

properties = {
    "Predictive Verb": (
        "Keyword: Predictive",
        "Keyword: non-Predictive",
        "Keyword: Predictive",
    ),
    "Uncertain Term Present": (
        "Keyword: Modal Verbs",
        "Keyword: non-Modal Verbs",
        "Keyword: Modal Verbs",
    ),
    "Time Period": (
        "Keyword: Time Expression",
        "Keyword: non-Time Expression",
        "Keyword: Time Expression",
    ),
}

row_labels = ["non-TOLSA", "TOLSA"]


def fired_count(dist_df, category):
    names = dist_df.iloc[:, 0].astype(str).str.strip()
    matched = dist_df.loc[names == category]

    if matched.empty:
        raise ValueError(
            f"Category {category!r} not found. "
            f"Available categories: {names.tolist()}"
        )

    return pd.to_numeric(matched.iloc[0, 1])


def label_totals(seed):
    split_dir = base_dir / f"seed{seed}" / "in_domain" / "splits"

    labels = pd.concat([
        pd.read_csv(
            split_dir / f"x_y_{split}_set.csv",
            usecols=[LABEL_COLUMN],
        )
        for split in ("val", "test")
    ])[LABEL_COLUMN].astype(int)

    counts = labels.value_counts()
    return np.array([counts.get(0, 0), counts.get(1, 0)])


def seed_matrix(seed, embedding, category):
    dist_non = globals()[
        f"distribution_non_tolsam_seed{seed}_{embedding}"
    ]
    dist_tol = globals()[
        f"distribution_tolsam_seed{seed}_{embedding}"
    ]

    non_tol_total, tol_total = label_totals(seed)
    non_tol_hit = fired_count(dist_non, category)
    tol_hit = fired_count(dist_tol, category)

    matrix = np.array([
        [non_tol_total - non_tol_hit, non_tol_hit],
        [tol_total - tol_hit, tol_hit],
    ])

    if (matrix < 0).any():
        raise ValueError(
            f"Negative count for seed={seed}, embedding={embedding}, "
            f"category={category!r}."
        )

    return matrix


fig, axes = plt.subplots(1, len(properties), figsize=(18, 5))

for ax, (category, (title, neg_label, pos_label)) in zip(
    axes, properties.items()
):
    stack = np.stack([
        seed_matrix(seed, embedding, category)
        for seed in SEEDS
        for embedding in EMBEDDINGS
    ])

    mean = stack.mean(axis=0)
    std = stack.std(axis=0, ddof=1)

    annot = np.array([
        [
            f"{mean[i, j]:,.1f}\n± {std[i, j]:.1f}"
            for j in range(2)
        ]
        for i in range(2)
    ])

    sns.heatmap(
        mean,
        annot=annot,
        fmt="",
        cmap="Blues",
        linewidths=0.5,
        cbar=True,
        xticklabels=[neg_label, pos_label],
        yticklabels=row_labels,
        ax=ax,
    )

    ax.set_title("TOLSA x Glodd")
    ax.set_xlabel("Glodd")
    ax.set_ylabel("Ground Truth")
    plt.setp(ax.get_xticklabels(), rotation=360, va="center")
    ax.tick_params(axis="x", pad=7)
    plt.setp(ax.get_yticklabels(), rotation=90, va="center")
    ax.tick_params(axis="y", pad=7)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns

SEEDS = [3, 7, 33]
LABEL_COLUMN = "Ground Truth"

# distribution row name -> (panel title, rule-not-fired label, rule-fired label)
properties = {
    "Verb": (
        "Keyword: Predictive",
        "Keyword: non-Predictive",
        "Keyword: Predictive",
    ),  # VERBS list
    "Modal Verb": (
        "Keyword: Modal Verb",
        "Keyword: non-Modal Verb",
        "Keyword: Modal Verb",
    ),  # AUXILIARIES list
}

row_labels = ["non-TOLSA", "TOLSA"]


def fired_count(dist_df, category):
    """Exact match on the category name, so "Verb" never picks up "Modal Verb"."""
    names = dist_df.iloc[:, 0].astype(str).str.strip()
    return pd.to_numeric(dist_df.loc[names == category].iloc[0, 1])


def label_totals(seed):
    """[Non-TOLSA, TOLSA] row counts over val + test, read from the saved splits."""
    split_dir = base_dir / f"seed{seed}" / "in_domain" / "splits"
    labels = pd.concat(
        [pd.read_csv(split_dir / f"x_y_{split}_set.csv", usecols=[LABEL_COLUMN])
         for split in ("val", "test")]
    )[LABEL_COLUMN].astype(int)
    counts = labels.value_counts()
    return np.array([counts.get(0, 0), counts.get(1, 0)])


def seed_matrix(seed, category):
    """2x2 counts: rows = Actual (Non-TOLSA, TOLSA), columns = rule (not fired, fired)."""
    non_tol_total, tol_total = label_totals(seed)
    non_tol_hit = fired_count(globals()[f"distribution_non_tolsam_seed{seed}"], category)
    tol_hit = fired_count(globals()[f"distribution_tolsam_seed{seed}"], category)
    return np.array([
        [non_tol_total - non_tol_hit, non_tol_hit],
        [tol_total - tol_hit, tol_hit],
    ])


fig, axes = plt.subplots(1, 2, figsize=(14, 5))

for ax, (category, (title, neg_label, pos_label)) in zip(axes, properties.items()):
    stack = np.stack([seed_matrix(seed, category) for seed in SEEDS])
    mean = stack.mean(axis=0)
    std = stack.std(axis=0, ddof=1)

    annot = np.array([
        [f"{mean[i, j]:,.1f}\n± {std[i, j]:.1f}" for j in range(2)]
        for i in range(2)
    ])

    sns.heatmap(
        mean,
        annot=annot,
        fmt="",
        cmap="Blues",
        linewidths=0.5,
        cbar=True,
        xticklabels=[neg_label, pos_label],
        yticklabels=row_labels,
        ax=ax,
    )
    ax.set_title("TOLSA x CoNLL-2010")
    ax.set_xlabel("CoNLL-2010")
    ax.set_ylabel("Ground Truth")

    plt.setp(ax.get_xticklabels(), rotation=360, va="center")
    ax.tick_params(axis="x", pad=7)

    plt.setp(ax.get_yticklabels(), rotation=90, va="center")
    ax.tick_params(axis="y", pad=7)
plt.tight_layout()
plt.show()
```

## Confusion Matricies Combined

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns

SEEDS = [3, 7, 33]
EMBEDDINGS = ["spacy_small", "spacy_medium", "spacy_large"]
LABEL_COLUMN = "Ground Truth"
row_labels = ["non-TOLSA", "TOLSA"]


# Glodd: (lookup category, rule-not-fired label, rule-fired label)
glodd_properties = {
    "Predictive": (
        "Predictive Verb",
        "Keyword: non-Predictive",
        "Keyword: Predictive",
    ),
    "Modal Verb": (
        "Uncertain Term Present",
        "Keyword: non-Modal Verbs",
        "Keyword: Modal Verbs",
    ),
    "Time Expression": (
        "Time Period",
        "Keyword: non-Time Expression",
        "Keyword: Time Expression",
    ),
}

# CoNLL-2010: (lookup category, rule-not-fired label, rule-fired label)
conll_properties = {
    "Predictive": (
        "Verb",
        "Keyword: non-Predictive",
        "Keyword: Predictive",
    ),
    "Modal Verb": (
        "Modal Verb",
        "Keyword: non-Modal Verb",
        "Keyword: Modal Verb",
    ),
}


def fired_count(dist_df, category):
    names = dist_df.iloc[:, 0].astype(str).str.strip()
    matched = dist_df.loc[names == category]

    if matched.empty:
        raise ValueError(
            f"Category {category!r} not found. "
            f"Available categories: {names.tolist()}"
        )

    return pd.to_numeric(matched.iloc[0, 1])


def label_totals(seed):
    split_dir = base_dir / f"seed{seed}" / "in_domain" / "splits"

    labels = pd.concat([
        pd.read_csv(
            split_dir / f"x_y_{split}_set.csv",
            usecols=[LABEL_COLUMN],
        )
        for split in ("val", "test")
    ])[LABEL_COLUMN].astype(int)

    counts = labels.value_counts()
    return np.array([counts.get(0, 0), counts.get(1, 0)])


# Glodd: counts for one seed and embedding
def glodd_matrix(seed, embedding, category):
    dist_non = globals()[
        f"distribution_non_tolsam_seed{seed}_{embedding}"
    ]
    dist_tol = globals()[
        f"distribution_tolsam_seed{seed}_{embedding}"
    ]

    non_tol_total, tol_total = label_totals(seed)
    non_tol_hit = fired_count(dist_non, category)
    tol_hit = fired_count(dist_tol, category)

    matrix = np.array([
        [non_tol_total - non_tol_hit, non_tol_hit],
        [tol_total - tol_hit, tol_hit],
    ])

    if (matrix < 0).any():
        raise ValueError(
            f"Negative count for seed={seed}, embedding={embedding}, "
            f"category={category!r}."
        )

    return matrix


# CoNLL-2010: counts for one seed
def conll_matrix(seed, category):
    dist_non = globals()[f"distribution_non_tolsam_seed{seed}"]
    dist_tol = globals()[f"distribution_tolsam_seed{seed}"]

    non_tol_total, tol_total = label_totals(seed)
    non_tol_hit = fired_count(dist_non, category)
    tol_hit = fired_count(dist_tol, category)

    matrix = np.array([
        [non_tol_total - non_tol_hit, non_tol_hit],
        [tol_total - tol_hit, tol_hit],
    ])

    if (matrix < 0).any():
        raise ValueError(
            f"Negative count for seed={seed}, category={category!r}."
        )

    return matrix


# Plot mean counts with sample standard deviation annotations
def plot_heatmap(ax, matrices, title, xlabel, neg_label, pos_label):
    stack = np.stack(matrices)
    mean = stack.mean(axis=0)
    std = stack.std(axis=0, ddof=1)

    annot = np.array([
        [
            f"{mean[i, j]:,.1f}\n± {std[i, j]:.1f}"
            for j in range(2)
        ]
        for i in range(2)
    ])

    sns.heatmap(
        mean,
        annot=annot,
        fmt="",
        cmap="Blues",
        linewidths=0.5,
        cbar=True,
        xticklabels=[neg_label, pos_label],
        yticklabels=row_labels,
        ax=ax,
    )

    ax.set_title(title)
    ax.set_xlabel(xlabel)
    ax.set_ylabel("Ground Truth")

    plt.setp(ax.get_xticklabels(), rotation=360, va="center")
    ax.tick_params(axis="x", pad=7)
    plt.setp(ax.get_yticklabels(), rotation=90, va="center")
    ax.tick_params(axis="y", pad=7)


# ============================================================
# 2 x 3 figure
# ============================================================
fig, axes = plt.subplots(2, 3, figsize=(21, 10))


# Row 1, Column 1: chronicle2050
chronicle_matrix = np.array([
    [3163, 2089],
    [33, 1106],
])

sns.heatmap(
    chronicle_matrix,
    annot=True,
    fmt=",",
    cmap="Blues",
    linewidths=0.5,
    xticklabels=["non-Future-Related", "Future-Related"],
    yticklabels=row_labels,
    ax=axes[0, 0],
)

axes[0, 0].set_title("TOLSA × chronicle2050")
axes[0, 0].set_xlabel("Future-Related")
axes[0, 0].set_ylabel("Ground Truth")


# Row 1, Column 2: Glodd Predictive
category, neg_label, pos_label = glodd_properties["Predictive"]

plot_heatmap(
    axes[0, 1],
    [
        glodd_matrix(seed, embedding, category)
        for seed in SEEDS
        for embedding in EMBEDDINGS
    ],
    "TOLSA × Glodd: Predictive",
    "Glodd",
    neg_label,
    pos_label,
)


# Row 1, Column 3: CoNLL-2010 Predictive
category, neg_label, pos_label = conll_properties["Predictive"]

plot_heatmap(
    axes[0, 2],
    [
        conll_matrix(seed, category)
        for seed in SEEDS
    ],
    "TOLSA × CoNLL-2010: Predictive",
    "CoNLL-2010",
    neg_label,
    pos_label,
)


# Row 2, Column 1: Glodd Modal Verb
category, neg_label, pos_label = glodd_properties["Modal Verb"]

plot_heatmap(
    axes[1, 0],
    [
        glodd_matrix(seed, embedding, category)
        for seed in SEEDS
        for embedding in EMBEDDINGS
    ],
    "TOLSA × Glodd: Modal Verb",
    "Glodd",
    neg_label,
    pos_label,
)


# Row 2, Column 2: CoNLL-2010 Modal Verb
category, neg_label, pos_label = conll_properties["Modal Verb"]

plot_heatmap(
    axes[1, 1],
    [
        conll_matrix(seed, category)
        for seed in SEEDS
    ],
    "TOLSA × CoNLL-2010: Modal Verb",
    "CoNLL-2010",
    neg_label,
    pos_label,
)


# Row 2, Column 3: Glodd Time Expression
category, neg_label, pos_label = glodd_properties["Time Expression"]

plot_heatmap(
    axes[1, 2],
    [
        glodd_matrix(seed, embedding, category)
        for seed in SEEDS
        for embedding in EMBEDDINGS
    ],
    "TOLSA × Glodd: Time Expression",
    "Glodd",
    neg_label,
    pos_label,
)


plt.tight_layout()
plt.show()
```
