## Example in Introduction

```python
colors = {
    # Core TOLSA-M properties
    "Source": "#fc9ce7",   # pink
    "Target": "#fad0c4",   # peach
    "Date": "#08e76d",     # green
    "Outcome": "#a6c1ee",  # light blue

    # Additional annotations
    "Keyword": "#fc9ce7",
    "Past Tense": "#ff6b6b",
    "POS=Noun": "#ff6b6b",
    "Question": "#ffd166",
    "Dialogue Act": "#ffd166",
    "Domain": "#cdb4db",
}
```

```python
def create_visualization(text, entities, title, colors):
    doc_data = {
        "text": text,
        "ents": entities,
        "title": title
    }

    options = {"ents": list(colors.keys()), "colors": colors}

    # Render the visualization
    html = displacy.render(doc_data, style="ent", manual=True, jupyter=False, options=options)
    # Then in your function:
    display(HTML(html))
    # return html
```

```python
# ------------------------------------------------------------------
# Text Document 1
# ------------------------------------------------------------------

title = "Text Document 1: Missed Tense Marker"

text = ("Charles Barkley predicted the Knicks would win the "
        "Eastern Conference Finals.")

entities = [
    {"start": 0, "end": 15, "label": "Source"},
    {"start": 16, "end": 23, "label": "Keyword"},
    {"start": 23, "end": 25, "label": "Past Tense"},
    {"start": 30, "end": 37, "label": "Target"},
    {"start": 43, "end": 76, "label": "Outcome"},
]

create_visualization(text, entities, title, colors)

entities = [
    {"start": 0, "end": 15, "label": "Source"},
    {"start": 30, "end": 37, "label": "Target"},
    {"start": 43, "end": 76, "label": "Outcome"},
]

create_visualization(text, entities, title, colors)
```
