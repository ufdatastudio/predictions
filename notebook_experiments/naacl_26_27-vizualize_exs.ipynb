## Example in Introduction

```python
colors = {
    # Core TOLSA-M properties
    "Target": "#F8E1CC",   # peach
    "Outcome":  "#F4ECC8",  # soft wheat
    "Source": "#FAE2F4",   # pink
    "Date": "#E8DDF7",     # green


    # Additional annotations
    "Keyword": "#fc9ce7",
    "Past Tense": "#ff6b6b",
    "POS=Noun": "#ff6b6b",
    "Question": "#ffd166",
    "Dialogue Act": "#ffd166",
    "Domain": "#cdb4db",
}
```

```python
def create_visualization(text, entities, title, colors):
    display(HTML(f"<div style='font-size:12px;font-weight:bold;margin-bottom:-3px'>{title}</div>"))

    doc_data = {
        "text": text,
        "ents": entities
    }

    options = {"ents": list(colors.keys()), "colors": colors}

    html = displacy.render(
        doc_data,
        style="ent",
        manual=True,
        jupyter=False,
        options=options
    )

    display(HTML(html))
```

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.patches import Patch

# =====================================================
# Individual Model Results
# =====================================================

models = [
    "Glodd-S", "Glodd-M", "Glodd-L",
    "CoNLL10",
    "spaCy-S", "spaCy-M", "spaCy-L",
    "ST-DR", "ST-ML12", "ST-ML6", "ST-MPNet",
    "LLM-ZS", "LLM-FS", "LLM-CoT",
    "BERT-Base", "BERT-Large"
]

# =====================================================
# Test Accuracy
# =====================================================

acc = [
    0.7109, 0.7134, 0.7186,
    0.7286,
    0.8336, 0.8506, 0.8818,
    0.9122, 0.9055, 0.8966, 0.7222,
    0.8393, 0.8257, 0.8235,
    0.9557, 0.9580
]

acc_std = [
    0.0050, 0.0052, 0.0042,
    0.0007,
    0.0162, 0.0163, 0.0522,
    0.0091, 0.0027, 0.0086, 0.0912,
    0.0179, 0.0232, 0.0805,
    0.0026, 0.0024
]

# =====================================================
# Macro F1
# =====================================================

f1 = [
    0.6912, 0.6943, 0.7005,
    0.7072,
    0.8290, 0.8407, 0.8705,
    0.9099, 0.9026, 0.8940, 0.6323,
    0.8286, 0.8392, 0.8087,
    0.9550, 0.9571
]

f1_std = [
    0.0087, 0.0087, 0.0069,
    0.0049,
    0.0504, 0.0285, 0.0992,
    0.0069, 0.0014, 0.0053, 0.1468,
    0.0269, 0.0143, 0.1114,
    0.0029, 0.0027
]

# =====================================================
# Colors
# =====================================================

colors = (
    ["#DD8452"] * 4 +  # Glodd-S/M/L + CoNLL10
    ["#4C72B0"] * 3 +  # spaCy
    ["#55A868"] * 4 +  # Sentence Transformer
    ["#C44E52"] * 3 +  # LLM
    ["#8172B2"] * 2    # BERT
)

x = np.arange(len(models))

# =====================================================
# Broken Axis Configuration
# =====================================================

GAP = 0.08

def add_y_axis_break(ax, y0, y1, width=0.006, n_zigs=3):
    ax.plot(
        [0, 0],
        [y0, y1],
        transform=ax.transAxes,
        color="white",
        linewidth=4,
        clip_on=False,
        zorder=10
    )

    ys = np.linspace(y0, y1, n_zigs + 2)
    xs = [0] + [
        (-width if i % 2 == 0 else width)
        for i in range(n_zigs)
    ] + [0]

    ax.plot(
        xs,
        ys,
        transform=ax.transAxes,
        color="black",
        linewidth=1,
        clip_on=False,
        zorder=11
    )

# =====================================================
# Plot Helper
# =====================================================

def plot_broken_panel(ax, values, stds, lower, ylabel, title, upper=1.0):

    def convert(v):
        return GAP + (np.asarray(v) - lower)

    heights = convert(values)

    bars = ax.bar(
        x,
        heights,
        yerr=stds,
        color=colors,
        edgecolor="black",
        linewidth=0.8,
        capsize=4
    )

    top = convert(upper)
    ax.set_ylim(0, top)

    ticks = np.round(
        np.arange(lower, upper + 1e-9, 0.1),
        1
    )

    ax.set_yticks([0] + list(convert(ticks)))
    ax.set_yticklabels(
        ["0"] + [f"{t:.1f}" for t in ticks]
    )

    ax.set_ylabel(ylabel, fontsize=11)
    ax.set_title(title, fontsize=12, weight="bold")

    ax.grid(
        axis="y",
        linestyle="--",
        alpha=0.35
    )

    for bar, value, height in zip(
        bars, values, heights
    ):
        ax.text(
            bar.get_x() + bar.get_width()/2,
            height + 0.01,
            f"{value:.3f}",
            ha="center",
            va="bottom",
            fontsize=7,
            rotation=90
        )

    add_y_axis_break(
        ax,
        y0=0,
        y1=GAP/top
    )

# =====================================================
# Create Figure
# =====================================================

fig, axes = plt.subplots(
    2,
    1,
    figsize=(16, 8.5),
    sharex=True
)

plot_broken_panel(
    axes[0],
    acc,
    acc_std,
    lower=0.60,
    ylabel="Test Accuracy",
    title="Test Accuracy by Model"
)

plot_broken_panel(
    axes[1],
    f1,
    f1_std,
    lower=0.40,
    ylabel="Macro F1",
    title="Macro F1 by Model"
)

# =====================================================
# X-Axis
# =====================================================

axes[1].set_xticks(x)
axes[1].set_xticklabels(
    models,
    rotation=45,
    ha="right",
    fontsize=10
)

# =====================================================
# Legend
# =====================================================

legend_handles = [
    Patch(
        facecolor="#DD8452",
        edgecolor="black",
        label="Inspired Baselines"
    ),
    Patch(
        facecolor="#4C72B0",
        edgecolor="black",
        label="Classical ML (spaCy Features)"
    ),
    Patch(
        facecolor="#55A868",
        edgecolor="black",
        label="Classical ML (Sentence Transformers)"
    ),
    Patch(
        facecolor="#C44E52",
        edgecolor="black",
        label="Prompted LLMs"
    ),
    Patch(
        facecolor="#8172B2",
        edgecolor="black",
        label="Fine-Tuned BERT"
    ),
]
fig.legend(
    handles=legend_handles,
    loc="upper center",
    bbox_to_anchor=(0.5, 1.03),
    ncol=5,
    frameon=False,
    fontsize=10
)

# =====================================================
# Formatting
# =====================================================

plt.tight_layout(
    rect=[0, 0, 1, 0.95]
)

# plt.savefig(
#     "rq4_model_comparison.png",
#     dpi=300,
#     bbox_inches="tight"
# )

plt.show()
```
