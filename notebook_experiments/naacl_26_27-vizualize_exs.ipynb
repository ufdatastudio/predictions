## Example in Introduction

```python
colors = {
    # Core TOLSA-M properties
    "Source": "#fc9ce7",   # pink
    "Target": "#fad0c4",   # peach
    "Date": "#08e76d",     # green
    "Outcome": "#a6c1ee",  # light blue

    # Additional annotations
    "Keyword": "#fc9ce7",
    "Past Tense": "#ff6b6b",
    "POS=Noun": "#ff6b6b",
    "Question": "#ffd166",
    "Dialogue Act": "#ffd166",
    "Domain": "#cdb4db",
}
```

```python
def create_visualization(text, entities, title, colors):
    display(HTML(f"<div style='font-size:12px;font-weight:bold;margin-bottom:-3px'>{title}</div>"))

    doc_data = {
        "text": text,
        "ents": entities
    }

    options = {"ents": list(colors.keys()), "colors": colors}

    html = displacy.render(
        doc_data,
        style="ent",
        manual=True,
        jupyter=False,
        options=options
    )

    display(HTML(html))
```

# Results: Task 1

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.patches import Patch

# =====================================================
# Data from Table RQ4
# =====================================================

models = [
    "spaCy-S", "spaCy-M", "spaCy-L",
    "ST-DR", "ST-ML12", "ST-ML6", "ST-MPNet",
    "LLM-ZS", "LLM-FS", "LLM-CoT"
]

# Test Accuracy
acc = [
    0.8336, 0.8506, 0.8818,
    0.9122, 0.9055, 0.8966, 0.7222,
    0.8393, 0.8257, 0.8235
]

acc_std = [
    0.0162, 0.0163, 0.0522,
    0.0091, 0.0027, 0.0086, 0.0912,
    0.0179, 0.0232, 0.0805
]

# F1
f1 = [
    0.8290, 0.8407, 0.8705,
    0.9099, 0.9026, 0.8940, 0.6323,
    0.8286, 0.8392, 0.8087
]

f1_std = [
    0.0504, 0.0285, 0.0992,
    0.0069, 0.0014, 0.0053, 0.1468,
    0.0269, 0.0143, 0.1114
]

# =====================================================
# Colors
# =====================================================

colors = (
    ["#4C72B0"] * 3 +  # spaCy
    ["#55A868"] * 4 +  # Sentence Transformers
    ["#C44E52"] * 3    # LLM Prompting
)

x = np.arange(len(models))

# =====================================================
# Figure
# =====================================================

fig, axes = plt.subplots(
    nrows=2,
    ncols=1,
    figsize=(12, 8),
    sharex=True
)

# =====================================================
# Test Accuracy
# =====================================================

bars_acc = axes[0].bar(
    x,
    acc,
    yerr=acc_std,
    color=colors,
    edgecolor="black",
    capsize=5
)

axes[0].set_ylabel("Test Accuracy")
axes[0].set_ylim(0.60, 1.00)
axes[0].set_title("Test Accuracy")

for bar, value in zip(bars_acc, acc):
    axes[0].text(
        bar.get_x() + bar.get_width() / 2,
        value + 0.01,
        f"{value:.3f}",
        ha="center",
        fontsize=8
    )

# =====================================================
# F1
# =====================================================

bars_f1 = axes[1].bar(
    x,
    f1,            # <- CORRECT F1 means
    yerr=f1_std,   # <- CORRECT F1 std
    color=colors,
    edgecolor="black",
    capsize=5
)

axes[1].set_ylabel("F1")
axes[1].set_ylim(0.40, 1.00)
axes[1].set_title("F1")

for bar, value in zip(bars_f1, f1):
    axes[1].text(
        bar.get_x() + bar.get_width() / 2,
        value + 0.01,
        f"{value:.3f}",
        ha="center",
        fontsize=8
    )

# =====================================================
# Shared X-axis
# =====================================================

axes[1].set_xticks(x)
axes[1].set_xticklabels(models, rotation=45, ha="right")

# =====================================================
# Legend
# =====================================================

legend_handles = [
    Patch(facecolor="#4C72B0", edgecolor="black", label="spaCy"),
    Patch(facecolor="#55A868", edgecolor="black", label="Sentence Transformer"),
    Patch(facecolor="#C44E52", edgecolor="black", label="LLM Prompting"),
]

axes[0].legend(handles=legend_handles, loc="lower left")

# =====================================================
# Layout
# =====================================================

plt.tight_layout()

plt.savefig(
    "rq4_accuracy_f1_combined.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
